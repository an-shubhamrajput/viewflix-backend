### ⭐ CELL 4 — Convert Watchlist → Implicit Feedback

```python
def compute_implicit_feedback(row):
    score = 1.0
    if row["watched"] == 1:
        score += 2.0
    return score

joined_df["implicit_score"] = joined_df.apply(
    compute_implicit_feedback, axis=1
)

joined_df[["user_id", "movie_id", "watched", "implicit_score"]].head()
```

### CELL 5 — Identify Genre Columns

```python
genre_cols = [c for c in movies_df.columns if c.startswith("genre_")]
len(genre_cols), genre_cols[:10]
```

### CELL 6 — Build User Profiles

```python
def build_user_profiles(joined_df, genre_cols):
    profiles = []

    for user_id, group in joined_df.groupby("user_id"):
        weights = group["implicit_score"].values.reshape(-1, 1)

        genre_matrix = group[genre_cols].values
        genre_pref = (genre_matrix * weights).sum(axis=0)
        genre_pref = genre_pref / (genre_pref.sum() + 1e-6)

        profile = {
            "user_id": user_id,
            "popularity_pref": np.average(
                group["popularity_norm"], weights=group["implicit_score"]
            ),
            "recency_pref": np.average(
                group["recency_score"], weights=group["implicit_score"]
            ),
        }

        for i, col in enumerate(genre_cols):
            profile[col] = genre_pref[i]

        profiles.append(profile)

    return pd.DataFrame(profiles)


profiles_df = build_user_profiles(joined_df, genre_cols)
profiles_df.head()
```

### CELL 7 — Hybrid Recommendation Function

```python
import numpy as np

def recommend_for_user(
    user_profile,
    movies_df,
    genre_cols,
    watchlist_df,
    top_k=20,
):
    """
    Returns top_k movie recommendations for a user
    - Scores by genre, popularity, recency
    - Removes watched movies
    - Deduplicates same movie across multiple source_db
    """

    # ----------------------------
    # 1. GENRE SCORE
    # ----------------------------
    user_genres = user_profile[genre_cols].values.astype(float)
    movie_genres = movies_df[genre_cols].values.astype(float)

    genre_score = movie_genres @ user_genres

    # ----------------------------
    # 2. POPULARITY SCORE
    # ----------------------------
    popularity_score = 1 - np.abs(
        movies_df["popularity_norm"].values
        - user_profile["popularity_pref"]
    )

    # ----------------------------
    # 3. RECENCY SCORE
    # ----------------------------
    recency_score = 1 - np.abs(
        movies_df["recency_score"].values
        - user_profile["recency_pref"]
    )

    # ----------------------------
    # 4. FINAL WEIGHTED SCORE
    # ----------------------------
    final_score = (
        0.6 * genre_score
        + 0.25 * popularity_score
        + 0.15 * recency_score
    )

    # ----------------------------
    # 5. ATTACH SCORE
    # ----------------------------
    result = movies_df.copy()
    result["score"] = final_score

    # ----------------------------
    # 6. REMOVE WATCHED MOVIES
    # ----------------------------
    watched_movies = set(
        watchlist_df[
            watchlist_df["user_id"] == user_profile["user_id"]
        ]["movie_id"]
    )

    result = result[~result["movie_id"].isin(watched_movies)]

    # ----------------------------
    # 7. SORT BY SCORE
    # ----------------------------
    result = result.sort_values("score", ascending=False)

    # ----------------------------
    # 8. DEDUPLICATE SAME MOVIE
    #    (keep best-scoring version)
    # ----------------------------
    result = result.drop_duplicates(
        subset="title",
        keep="first"
    )

    # ----------------------------
    # 9. RETURN TOP-K
    # ----------------------------
    return result.head(top_k)[
        ["movie_id", "title", "score", "source_db"]
    ]
```

### 🎬 CELL 8 — Get Recommendations for a User

```python
# Pick a user
user_profile = profiles_df.iloc[1]

recommendations = recommend_for_user(
    user_profile=user_profile,
    movies_df=movies_df,
    genre_cols=genre_cols,
    watchlist_df=watchlist_df,
    top_k=40,
)

recommendations
```
